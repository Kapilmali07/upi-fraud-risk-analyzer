```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("C:/Users/malik/upi-fraud-risk-analyzer/data/PS_20174392719_1491204439457_log.csv")
```

```python
fraud_by_type = df[df['isFraud']==1]['type'].value_counts()
```

```python
fraud_by_type
```

```python
fraud_by_type.plot(kind='bar', color='Red', figsize=(8,4))
plt.title('Fraud count by transaction type')
plt.xlabel('Transaction type')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('C:/Users/malik/upi-fraud-risk-analyzer//outputs/fraud_by_type.png')
plt.show()
```

```python
fraud_hourly = df[df['isFraud']==1]['step'].value_counts().sort_index()
```

```python
fraud_hourly
```

```python
fraud_hourly.plot(kind='line', color='Red', figsize=(8,4))
plt.title('Fraud transaction by hours')
plt.xlabel('Hours (step)')
plt.ylabel('Fraud Count')
plt.tight_layout()
plt.savefig('C:/Users/malik/upi-fraud-risk-analyzer//outputs/fraud_by_hour.png')
plt.show()
```

```python
normal_amt = df[df['isFraud'] == 0]['amount'].clip(upper=500000)

fraud_amt = df[df['isFraud'] == 1]['amount']

plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
plt.hist(normal_amt, color='Green', bins=50)
plt.ylabel('Frequency')
plt.title("Normal Transactions")


plt.subplot(1,2,2)
plt.hist(fraud_amt, color='Red', bins=50)
plt.ylabel('Frequency')
plt.title("Fraud Transactions")

plt.tight_layout()
plt.savefig('C:/Users/malik/upi-fraud-risk-analyzer//outputs/amount_distribution.png')
plt.show()
```
