```python
from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch
from reportlab.platypus import (SimpleDocTemplate, Paragraph, Spacer, 
                                 Table, TableStyle, HRFlowable)
from reportlab.lib.enums import TA_CENTER, TA_LEFT

# Output path
output_path = 'C:/Users/malik/upi-fraud-risk-analyzer/outputs/upi_fraud_risk_report.pdf'

doc = SimpleDocTemplate(
    output_path,
    pagesize=A4,
    rightMargin=50, leftMargin=50,
    topMargin=50, bottomMargin=50
)

styles = getSampleStyleSheet()
elements = []

# ── Custom styles ──────────────────────────────────────────
title_style = ParagraphStyle(
    'Title', fontSize=20, fontName='Helvetica-Bold',
    textColor=colors.HexColor('#185FA5'),
    alignment=TA_CENTER, spaceAfter=6
)
subtitle_style = ParagraphStyle(
    'Subtitle', fontSize=11, fontName='Helvetica',
    textColor=colors.HexColor('#444441'),
    alignment=TA_CENTER, spaceAfter=4
)
section_style = ParagraphStyle(
    'Section', fontSize=13, fontName='Helvetica-Bold',
    textColor=colors.HexColor('#185FA5'),
    spaceBefore=14, spaceAfter=6,
    borderPad=4
)
body_style = ParagraphStyle(
    'Body', fontSize=10, fontName='Helvetica',
    textColor=colors.HexColor('#2C2C2A'),
    leading=16, spaceAfter=6
)
bold_style = ParagraphStyle(
    'Bold', fontSize=10, fontName='Helvetica-Bold',
    textColor=colors.HexColor('#2C2C2A'),
    leading=16, spaceAfter=4
)

# ── PAGE 1 — Project Overview ──────────────────────────────
elements.append(Spacer(1, 0.3*inch))
elements.append(Paragraph("UPI Transaction Risk Profiler", title_style))
elements.append(Paragraph("Fraud Detection & Behavioral Risk Analysis", subtitle_style))
elements.append(Paragraph("Python · MySQL · Power BI · ReportLab", subtitle_style))
elements.append(Spacer(1, 0.1*inch))
elements.append(HRFlowable(width="100%", thickness=1.5,
                            color=colors.HexColor('#185FA5')))
elements.append(Spacer(1, 0.2*inch))

# Business Problem
elements.append(Paragraph("Business Problem", section_style))
elements.append(Paragraph(
    "UPI processed 17,220 crore transactions worth ₹246 lakh crore in FY2024-25, "
    "making India the world's largest real-time payments market by volume. With fraud "
    "cases rising, banks need automated behavioral risk detection systems that go "
    "beyond static rules. This project builds a Fraud Velocity Index (FVI) to detect "
    "anomalous UPI payment behavior and flag high-risk transactions before loss occurs.",
    body_style))

elements.append(Spacer(1, 0.1*inch))

# Dataset Overview Table
elements.append(Paragraph("Dataset Overview", section_style))
dataset_data = [
    ['Property', 'Detail'],
    ['Source', 'PaySim Synthetic Financial Dataset (Kaggle)'],
    ['Total Size', '6.3 million transactions, 11 features'],
    ['Fraud Rate', '0.13% — 8,213 fraudulent transactions'],
    ['Fraud Types', 'TRANSFER and CASH_OUT only'],
    ['Sample Used', '50,000 rows loaded into MySQL & Power BI'],
]
dataset_table = Table(dataset_data, colWidths=[2*inch, 4.2*inch])
dataset_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#185FA5')),
    ('TEXTCOLOR', (0,0), (-1,0), colors.white),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,-1), 9),
    ('FONTNAME', (0,1), (0,-1), 'Helvetica-Bold'),
    ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#F5F8FC')),
    ('ROWBACKGROUNDS', (0,1), (-1,-1),
     [colors.HexColor('#F5F8FC'), colors.white]),
    ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#B5D4F4')),
    ('PADDING', (0,0), (-1,-1), 7),
    ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
]))
elements.append(dataset_table)
elements.append(Spacer(1, 0.2*inch))

# Tools Table
elements.append(Paragraph("Tools & Technologies", section_style))
tools_data = [
    ['Tool', 'Purpose'],
    ['Python', 'EDA, feature engineering, visualization'],
    ['MySQL', 'Database design, SQL analysis queries'],
    ['Power BI', 'Interactive dashboard, What-If simulator'],
    ['ReportLab', 'PDF summary report generation'],
    ['GitHub', 'Version control, project documentation'],
]
tools_table = Table(tools_data, colWidths=[2*inch, 4.2*inch])
tools_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#185FA5')),
    ('TEXTCOLOR', (0,0), (-1,0), colors.white),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,-1), 9),
    ('FONTNAME', (0,1), (0,-1), 'Helvetica-Bold'),
    ('ROWBACKGROUNDS', (0,1), (-1,-1),
     [colors.HexColor('#F5F8FC'), colors.white]),
    ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#B5D4F4')),
    ('PADDING', (0,0), (-1,-1), 7),
    ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
]))
elements.append(tools_table)

# ── PAGE 2 — Key Findings ──────────────────────────────────
from reportlab.platypus import PageBreak
elements.append(PageBreak())

elements.append(Paragraph("Fraud Velocity Index (FVI)", section_style))
elements.append(Paragraph(
    "FVI is a custom behavioral risk score created for this project, "
    "combining three transaction signals into a single risk number (0–6 scale).",
    body_style))

fvi_data = [
    ['Signal', 'Logic', 'Weight'],
    ['Balance Mismatch', 'Account balances do not reconcile', '2 points'],
    ['Zero Drain', 'Account wiped to zero after transaction', '3 points'],
    ['Amount Ratio', 'Transaction large relative to account size', '0–1 points'],
]
fvi_table = Table(fvi_data, colWidths=[1.8*inch, 2.8*inch, 1.6*inch])
fvi_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#185FA5')),
    ('TEXTCOLOR', (0,0), (-1,0), colors.white),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,-1), 9),
    ('ROWBACKGROUNDS', (0,1), (-1,-1),
     [colors.HexColor('#F5F8FC'), colors.white]),
    ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#B5D4F4')),
    ('PADDING', (0,0), (-1,-1), 7),
    ('ALIGN', (2,0), (2,-1), 'CENTER'),
]))
elements.append(fvi_table)
elements.append(Spacer(1, 0.2*inch))

elements.append(Paragraph("Key Findings", section_style))

findings = [
    ("Finding 1 — Fraud concentrated in 2 transaction types",
     "100% of fraud occurs in TRANSFER (0.84% rate) and CASH_OUT (0.20% rate). "
     "PAYMENT, DEBIT, and CASH_IN show zero fraud. Risk monitoring can focus "
     "exclusively on these two types, reducing screening cost by 60%+."),
    ("Finding 2 — Fraud peaks at specific hours",
     "Fraud spikes around step 300 with 3x higher volume than average hours. "
     "Automated monitoring and alerts should be strongest during this peak window "
     "when human oversight is lowest."),
    ("Finding 3 — FVI successfully separates fraud from normal",
     "Critical Risk segment (FVI ≥ 4.0) contains 12,012 transactions and captures "
     "95.7% of all actual fraud. Low Risk segment (FVI < 1.5) contains 7,305 "
     "transactions with zero fraud — these can be cleared instantly with no review."),
    ("Finding 4 — High value, single-use fraud accounts",
     "Top fraud accounts each transferred exactly ₹1 crore in a single transaction. "
     "Every fraud account was used exactly once, making traditional blacklisting "
     "ineffective. Behavioral scoring is the correct detection approach."),
]

for title, body in findings:
    elements.append(Paragraph(title, bold_style))
    elements.append(Paragraph(body, body_style))
    elements.append(Spacer(1, 0.08*inch))

# FVI Results Table
elements.append(Spacer(1, 0.1*inch))
elements.append(Paragraph("FVI Risk Segmentation Results", section_style))
seg_data = [
    ['Risk Segment', 'Transactions', 'Actual Fraud', 'Fraud Rate'],
    ['Critical Risk (FVI ≥ 4)', '12,012', '67', '0.56%'],
    ['High Risk (FVI 3–4)', '20,134', '1', '0.00%'],
    ['Medium Risk (FVI 1.5–3)', '10,549', '2', '0.02%'],
    ['Low Risk (FVI < 1.5)', '7,305', '0', '0.00%'],
]
seg_table = Table(seg_data, colWidths=[2.2*inch, 1.5*inch, 1.5*inch, 1*inch])
seg_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#185FA5')),
    ('TEXTCOLOR', (0,0), (-1,0), colors.white),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,-1), 9),
    ('FONTNAME', (0,1), (0,-1), 'Helvetica-Bold'),
    ('BACKGROUND', (0,1), (-1,1), colors.HexColor('#FCEBEB')),
    ('ROWBACKGROUNDS', (0,2), (-1,-1),
     [colors.HexColor('#F5F8FC'), colors.white]),
    ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#B5D4F4')),
    ('PADDING', (0,0), (-1,-1), 7),
    ('ALIGN', (1,0), (-1,-1), 'CENTER'),
]))
elements.append(seg_table)

# ── PAGE 3 — Recommendations ──────────────────────────────
elements.append(PageBreak())

elements.append(Paragraph("Business Recommendations", section_style))
elements.append(HRFlowable(width="100%", thickness=1,
                            color=colors.HexColor('#B5D4F4')))
elements.append(Spacer(1, 0.15*inch))

recommendations = [
    ("Recommendation 1 — Focus screening on TRANSFER and CASH_OUT",
     "Flag all TRANSFER and CASH_OUT transactions above ₹50,000 for enhanced "
     "automated screening. These are the only two types with confirmed fraud "
     "activity, allowing risk teams to eliminate unnecessary checks on 3 other "
     "transaction types entirely."),
    ("Recommendation 2 — Set FVI operational threshold at 4.0",
     "An FVI threshold of 4.0 captures 95.7% of fraud while limiting review "
     "workload to 24% of total transactions. This is the optimal balance between "
     "detection sensitivity and operational cost. The Power BI What-If simulator "
     "allows risk managers to adjust this threshold in real time as fraud patterns "
     "evolve."),
    ("Recommendation 3 — Heighten monitoring during peak fraud hours",
     "Implement enhanced automated alerts during step 280–320 (equivalent to peak "
     "fraud window). Fraud volume during this period is 3x above daily average. "
     "Deploying additional rule-based checks during this window can intercept "
     "fraud before settlement occurs."),
    ("Recommendation 4 — Replace blacklisting with behavioral scoring",
     "Since every fraud account in this dataset was used exactly once, account-level "
     "blacklisting has zero effectiveness against new fraud attempts. FVI-based "
     "transaction-level behavioral scoring is the correct long-term approach, "
     "consistent with RBI's 2024 revised guidelines on early warning systems."),
]

for i, (title, body) in enumerate(recommendations):
    rec_data = [[Paragraph(title, bold_style)],
                [Paragraph(body, body_style)]]
    rec_table = Table(rec_data, colWidths=[6.2*inch])
    rec_table.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#E6F1FB')),
        ('BACKGROUND', (0,1), (-1,1), colors.white),
        ('BOX', (0,0), (-1,-1), 0.5, colors.HexColor('#185FA5')),
        ('PADDING', (0,0), (-1,-1), 10),
        ('LEFTPADDING', (0,0), (-1,-1), 12),
    ]))
    elements.append(rec_table)
    elements.append(Spacer(1, 0.15*inch))

# Footer
elements.append(Spacer(1, 0.2*inch))
elements.append(HRFlowable(width="100%", thickness=1,
                            color=colors.HexColor('#B5D4F4')))
elements.append(Spacer(1, 0.1*inch))
elements.append(Paragraph(
    "Kapil Mali · Data Analyst · Kantar Analytics, Pune · "
    "github.com/Kapilmali07/upi-fraud-risk-analyzer",
    ParagraphStyle('Footer', fontSize=8, fontName='Helvetica',
                   textColor=colors.HexColor('#888780'),
                   alignment=TA_CENTER)))

# ── Build PDF ──────────────────────────────────────────────
doc.build(elements)
print("PDF created successfully!")
print(f"Saved to: {output_path}")
```
