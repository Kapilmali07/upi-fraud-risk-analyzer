```python
import pandas as pd
from sqlalchemy import create_engine

# Load engineered dataset
df = pd.read_csv('C:/Users/malik/upi-fraud-risk-analyzer/data/upi_engineered.csv')
print("Full dataset shape:", df.shape)

# Take 50K sample for MySQL
sample = df.sample(n=50000, random_state=42)
print("Sample shape:", sample.shape)
```

```python
# Rename columns to match MySQL table
sample = sample.rename(columns={
    'step'            : 'step',
    'type'            : 'txn_type',
    'amount'          : 'amount',
    'nameOrig'        : 'name_orig',
    'oldbalanceOrg'   : 'old_bal_orig',
    'newbalanceOrig'  : 'new_bal_orig',
    'nameDest'        : 'name_dest',
    'oldbalanceDest'  : 'old_bal_dest',
    'newbalanceDest'  : 'new_bal_dest',
    'isFraud'         : 'is_fraud'
})

# Keep only needed columns
cols = ['step','txn_type','amount','name_orig','old_bal_orig',
        'new_bal_orig','name_dest','old_bal_dest','new_bal_dest',
        'is_fraud','fvi','zero_drain','mismatch_flag']

sample = sample[cols]
print("Final columns:", list(sample.columns))
```

```python
engine = create_engine('mysql+pymysql://root:Kapil@localhost/upi_fraud_db')

# Load into MySQL
sample.to_sql('transactions', con=engine, if_exists='replace', index=False)

print("Done! Rows loaded: 50,000")
```
